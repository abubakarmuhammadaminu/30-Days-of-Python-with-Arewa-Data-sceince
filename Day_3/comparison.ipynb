```python
# The slope and intercepts of y = 2x + 2
# For linear equation in slope-intercepts form (y = mx + b):
# Slope(m) = 2
# y-intercept(x = 0) = -2 (point: (0, -2))
# x-intercept (y = 0): 0 = 2x - 2
# x = 1 (point: (1, 0))
slope_1 = 2
x_intercept = 1
y_intercept = -2

print(f"Slope is ",slope_1)
print(f"X-intercept is ", x_intercept)
print(f"Y-intercept is ", y_intercept)
```

```python
x_1 = 2**1
y_1 = 2**1
x_2 = 6**2
y_2 = 10**2

slope = (y_2 - y_1 / x_2 - x_1)
print(f"The slope is", slope)
```

```python
print(slope == slope_1)
print(slope > slope_1)
print(slope < slope_1)
print(slope >= slope_1)
print(slope <= slope_1)
print(slope != slope_1)
```
